# Save number of LUTS after yosys synthesis

This notebook reads the synthesis log file generated by yosys and extracts the number of LUTs used in the design. The number of LUTs is saved in a .json file for further analysis.

```python
import os
import argparse
import sys

print(os.getcwd())
os.chdir('../')
print(os.getcwd())
sys.path.append(os.path.join(os.path.dirname('app'), "./src"))
from src.ultils.runScripts import RunScripts
from src.ultils.style import Mensage
```

```python
args = {
    'name': 'minist_treelut',
    'output': 'results', # defaut
    'verilog': 'misc/data/TreeLUT/verilog/*.v',
    'top_module': 'TreeLUT',
    'lut_size': 5, # default
    'debug': True
}
```

```python
# Ensure output directories exist
PATH = f'results/{args['name']}' if args['output'] == 'results' else os.path.join(args['output'], args['name'])
os.makedirs(PATH, exist_ok=True)

# Move verilog file to the correct path
os.makedirs(os.path.join(PATH, 'verilog'), exist_ok=True)
os.system(f"cp {args['verilog']} {os.path.join(PATH, 'verilog')}")
```

```python
mensage = Mensage()
mensage.title()
    
# Create RunScripts instance
scripts = RunScripts(
    path=PATH,
    verilog_path=f'{os.path.basename(args['verilog'])}', 
    top_module=args['top_module'],
    args={'otimizations': ['compress2rs'], 'lut_k': args['lut_size']},
    debug=args['debug']
)
scripts.run_yosys()
scripts.run_abc()
```
