```python
import zipfile

print(os.getcwd())
# os.chdir('app')
print(os.getcwd())

zip_path = "misc/data/susy_combinacoes.zip"  # Substitua pelo caminho do seu arquivo ZIP
extract_to = "misc/data"  # Substitua pelo diretório onde deseja extrair os arquivos

with zipfile.ZipFile(zip_path, 'r') as zip_ref:
    zip_ref.extractall(extract_to)

print("Arquivos extraídos com sucesso!")
```

```python
args = {
    'name': 'ADULTS_treelut',
    'output': 'results', # defaut
    'verilog': 'misc/data/Adult/TreeLUT/verilog/*.v',
    'top_module': 'TreeLUT',
    'lut_size': 5, # default
    'debug': True
}
```

```python
# Ensure output directories exist
os.chdir('../')
print(os.getcwd())
PATH = f'results/{args['name']}' if args['output'] == 'results' else os.path.join(args['output'], args['name'])
os.makedirs(PATH, exist_ok=True)

# Move verilog file to the correct path
os.makedirs(os.path.join(PATH, 'verilog'), exist_ok=True)
os.system(f"cp {args['verilog']} {os.path.join(PATH, 'verilog')}")
```

```python
mensage = Mensage()
mensage.title()
    
# Create RunScripts instance
scripts = RunScripts(
    path=PATH,
    verilog_path=f'{os.path.basename(args['verilog'])}', 
    top_module=args['top_module'],
    args={'otimizations': ['compress2rs'], 'lut_k': args['lut_size']},
    debug=args['debug']
)
scripts.run_yosys()
scripts.run_abc()
```

```python

# Loop principal para todas as combinações
for w_feature in [1, 2, 3, 4, 5]:
    for w_tree in [1, 2, 3, 4, 5]:
        combo_dir = f'w_feature_{w_feature}_w_tree_{w_tree}'
        verilog_source = f'misc/data/SUSY/{combo_dir}/TreeLUT/verilog/*.v'

        args = {
        'name': 'SUSY_treelut',
        'output': 'results', # defaut
        'verilog': verilog_source,
        'top_module': 'TreeLUT',
        'lut_size': 5, # default
        'debug': True
        }
    
        # Ensure output directories exist
        # os.chdir('app')
        print(os.getcwd())
        PATH = f'results/{args['name']}' if args['output'] == 'results' else os.path.join(args['output'], args['name'])
        os.makedirs(PATH, exist_ok=True)
        
        # Move verilog file to the correct path
        os.makedirs(os.path.join(PATH, 'verilog'), exist_ok=True)
        os.system(f"cp {args['verilog']} {os.path.join(PATH, 'verilog')}")
    
        
        mensage = Mensage()
        mensage.title()
            
        # Create RunScripts instance
        scripts = RunScripts(
            path=PATH,
            verilog_path=f'{os.path.basename(args['verilog'])}', 
            top_module=args['top_module'],
            args={'otimizations': ['compress2rs'], 'lut_k': args['lut_size']},
            debug=args['debug']
        )
        print(f'w_feature_{w_feature}_w_tree_{w_tree}')
        scripts.run_yosys()
        scripts.run_abc()
    
```
