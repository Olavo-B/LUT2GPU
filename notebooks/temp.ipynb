```python
def generate_testbench(num_inputs, num_outputs, module_name="TreeLUT", filename="generated_testbench.v"):
    with open(filename, "w") as f:
        f.write("`timescale 1ns/1ps\n\n")
        f.write("module sim ();\n\n")
        f.write("""reg [3135:0] x_test [0:9999];
reg [3:0] y_test [0:9999];

reg [3135:0] treelut_input;
wire [49:0] treelut_output;

wire [3:0] out_predicted;
reg [3:0] out_expected;

integer i, j;
reg [13:0] result_false;
reg [13:0] result_true;
reg clk;""")
        f.write(f"\n\n{module_name} {module_name}_inst (\n")
        # Writing input connections
        for i in range(num_inputs):
            comma = "," if i < num_inputs - 1 or num_outputs > 0 else ""
            f.write(f"    .i{i:02d} (treelut_input[{i}]){comma}\n")
        
        # Writing output connections
        for i in range(num_outputs):
            comma = "," if i < num_outputs - 1 else ""
            f.write(f"    .o{i:02d} (treelut_output[{i}]){comma}\n")
        
        f.write(");\n\n")
        
        f.write("""argmax argmax_inst(treelut_output, out_predicted);

initial begin
	
	result_false <= 0; result_true <= 0;
	$readmemb("X_test.mem", x_test); $readmemb("y_test.mem", y_test);
end

initial begin
	clk <= 0;
	forever #5 clk <= ~clk;
end

initial begin
	#105
	for (i = 0; i <= 10000; i = i + 1) begin
		#10
		treelut_input <= x_test[i];
	end
end

initial begin
	#125
	for (j = 0; j <= 10000; j = j + 1) begin
		#10
		out_expected <= y_test[j];
		if(out_predicted == out_expected)
			result_true <= result_true + 1;
		else
			result_false <= result_false+1;
            $display("Sample: %d: %d/%d", j, result_true, result_false);
	end
	$finish;
end
endmodule

module argmax(input wire [49:0] i, output reg [3:0] o);

wire [4:0] sum [0:9];
assign sum[0] = i[4:0];
assign sum[1] = i[9:5];
assign sum[2] = i[14:10];
assign sum[3] = i[19:15];
assign sum[4] = i[24:20];
assign sum[5] = i[29:25];
assign sum[6] = i[34:30];
assign sum[7] = i[39:35];
assign sum[8] = i[44:40];
assign sum[9] = i[49:45];
always @(*) begin
	if((sum[0] >= sum[1]) & (sum[0] >= sum[2]) & (sum[0] >= sum[3]) & (sum[0] >= sum[4]) & (sum[0] >= sum[5]) & (sum[0] >= sum[6]) & (sum[0] >= sum[7]) & (sum[0] >= sum[8]) & (sum[0] >= sum[9]))
		o = 4'd0;
	else if((sum[1] >= sum[2]) & (sum[1] >= sum[3]) & (sum[1] >= sum[4]) & (sum[1] >= sum[5]) & (sum[1] >= sum[6]) & (sum[1] >= sum[7]) & (sum[1] >= sum[8]) & (sum[1] >= sum[9]))
		o = 4'd1;
	else if((sum[2] >= sum[3]) & (sum[2] >= sum[4]) & (sum[2] >= sum[5]) & (sum[2] >= sum[6]) & (sum[2] >= sum[7]) & (sum[2] >= sum[8]) & (sum[2] >= sum[9]))
		o = 4'd2;
	else if((sum[3] >= sum[4]) & (sum[3] >= sum[5]) & (sum[3] >= sum[6]) & (sum[3] >= sum[7]) & (sum[3] >= sum[8]) & (sum[3] >= sum[9]))
		o = 4'd3;
	else if((sum[4] >= sum[5]) & (sum[4] >= sum[6]) & (sum[4] >= sum[7]) & (sum[4] >= sum[8]) & (sum[4] >= sum[9]))
		o = 4'd4;
	else if((sum[5] >= sum[6]) & (sum[5] >= sum[7]) & (sum[5] >= sum[8]) & (sum[5] >= sum[9]))
		o = 4'd5;
	else if((sum[6] >= sum[7]) & (sum[6] >= sum[8]) & (sum[6] >= sum[9]))
		o = 4'd6;
	else if((sum[7] >= sum[8]) & (sum[7] >= sum[9]))
		o = 4'd7;
	else if((sum[8] >= sum[9]))
		o = 4'd8;
	else
		o = 4'd9;
end

endmodule""")
    
    print(f"Testbench gerado e salvo em {filename}")

# Exemplo de uso
generate_testbench(num_inputs=3136, num_outputs=50, filename='misc/data/TreeLUT/testbench/testbench_copy.v')
```
